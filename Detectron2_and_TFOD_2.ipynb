Object center

Width & height

Aspect ratio

Position within image

Poor box alignment → poor IoU → weak training signal.

3️ Improves Model Accuracy (mAP & IoU)

Accurate annotations directly impact:

IoU (Intersection over Union)

Precision & Recall

mAP (mean Average Precision)

Noisy or incorrect labels:

Reduce precision

Increase false positives/negatives

Cause unstable training

Garbage annotations = garbage model.

4️ Helps Class Imbalance Handling

Proper labeling ensures:

All classes are represented

Rare objects are included

Model doesn’t become biased

Unbalanced annotations → biased predictions.

5️ Enables Generalization

High-quality, diverse annotations:

Different lighting

Different angles

Occlusions

Scale variations

This helps the model generalize to real-world scenarios.

Q3. What does batch size refer to in the context of model training?

Ans. Batch size is the number of training samples processed by a model before updating the model’s weights during one forward–backward pass.

In simple terms:

Batch size = How many data points the model looks at before it learns (updates).

How It Works

During training:

Model takes batch_size samples

Computes predictions

Calculates loss

Performs backpropagation

Updates weights

Then it moves to the next batch.

Example

If you have:

Dataset size = 1,000 images

Batch size = 100

Then:

Number of batches per epoch = 1000 / 100 = 10 batches

Model updates weights 10 times per epoch

Types of Gradient Descent (Based on Batch Size)
Type	Batch Size	Behavior
Batch Gradient Descent	Entire dataset	Stable but slow
Stochastic Gradient Descent (SGD)	1 sample	Very noisy but fast
Mini-batch Gradient Descent	8–512 (common)	Balanced & most practical

Most deep learning models use mini-batches.

Effect of Batch Size
🔹 Small Batch Size (e.g., 8–32)

More noisy gradients

Better generalization sometimes

Slower per epoch

Less memory usage

🔹 Large Batch Size (e.g., 256–1024)

Stable gradients

Faster computation on GPUs

Requires more memory

May generalize worse if too large

Q4. What is the purpose of pretrained weights in object detection models?

Ans. Pretrained weights are model parameters learned from training on a large dataset (e.g., ImageNet or COCO) that are reused to initialize a new model for a related task.

Instead of starting from random initialization, the model begins with already learned visual features.

Why They Are Important
1️ Faster Convergence

Random initialization → slow learning
Pretrained weights → model already understands:

Edges

Textures

Shapes

Basic object structures

Training becomes significantly faster.

2️ Better Performance (Higher mAP)

Large datasets like ImageNet contain millions of images.
Models pretrained on them learn strong feature representations.

When transferred to object detection:

Improves accuracy

Improves generalization

Reduces overfitting

Especially important when your dataset is small.

3️ Reduces Data Requirement

Training from scratch requires:

Huge labeled dataset

Massive compute

With pretrained weights:

Smaller dataset is sufficient

Less GPU time required

4️ Enables Transfer Learning

In object detection models (Faster R-CNN, YOLO, RetinaNet):

The backbone network (e.g., ResNet) is usually pretrained

The detection head is fine-tuned for the new dataset

Common strategies:

Freeze backbone layers

Fine-tune all layers

Partial fine-tuning

5️ Stabilizes Training

Random weights can cause:

Unstable gradients

Slow convergence

Poor local minima

Pretrained initialization gives the optimizer a strong starting point.

Technical Perspective

Pretrained weights provide:

Low-level feature extraction (edges, corners)

High-level semantic representations

Object detection models build on these features for:

Bounding box regression

 What Is a TFRecord?

A TFRecord file:

Stores data in binary format

Contains serialized tf.train.Example objects

Can store images + labels + bounding boxes together

Is optimized for streaming large datasets

Instead of loading thousands of individual image files, TFOD2 reads structured TFRecord files for faster training.

 What Does a TFRecord Contain (For Object Detection)?

Each record typically includes:

Encoded image (JPEG/PNG)

Image height & width

Class labels

Bounding box coordinates (xmin, ymin, xmax, ymax)

Optional masks or keypoints

Everything required for training is packed into one structured format.

 Why TFRecord Is Used in TFOD2
1️ Faster Data Loading

Binary format →

Faster disk I/O

Efficient sequential reading

Reduced file open/close overhead

This is critical for GPU training.

2️ Optimized for tf.data Pipeline

TFOD2 uses TensorFlow’s tf.data API:

Prefetching

Parallel reading

Caching

Shuffling

TFRecord integrates seamlessly into this pipeline.

3️ Scalable for Large Datasets

When training on:

COCO-scale datasets

Multi-GPU setups

Distributed training

TFRecord enables efficient data streaming without memory overload.

4️ Structured and Consistent Format

Ensures:

Uniform schema

Correct feature types

Q14. How do you perform inference with a trained Detectron2 model?

Ans. To run inference using Detectron2, you load the trained weights, configure the model, and pass an image through the predictor.

Below is the precise workflow.

 Step 1: Load Configuration and Weights
from detectron2.engine import DefaultPredictor
from detectron2.config import get_cfg
from detectron2 import model_zoo
import cv2

cfg = get_cfg()

# If using a model zoo config
cfg.merge_from_file(model_zoo.get_config_file(
    "COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml"
))

# If using your custom trained model
cfg.MODEL.WEIGHTS = "output/model_final.pth"

cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5  # confidence threshold
cfg.MODEL.DEVICE = "cuda"  # or "cpu"

predictor = DefaultPredictor(cfg)

 Step 2: Read the Image
image = cv2.imread("input.jpg")

 Step 3: Run Inference
outputs = predictor(image)


outputs contains:

pred_boxes

scores

pred_classes

pred_masks (if segmentation model)

 Step 4: Visualize Predictions
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog

v = Visualizer(
    image[:, :, ::-1],
    MetadataCatalog.get(cfg.DATASETS.TRAIN[0]),
    scale=1.0

out = v.draw_instance_predictions(outputs["instances"].to("cpu"))

cv2.imwrite("output.jpg", out.get_image()[:, :, ::-1])

Q16. What does fine-tuning pretrained weights involve?

Ans. Fine-tuning pretrained weights means taking a model that was previously trained on a large dataset (e.g., COCO or ImageNet) and adapting its learned parameters to a new, usually smaller, custom dataset.

Instead of training from scratch, you reuse learned feature representations and adjust them for your task.

 What Fine-Tuning Involves (Step-by-Step)
1️ Load Pretrained Weights

Initialize the model with weights trained on a large dataset.

Example:

Backbone pretrained on ImageNet

Detection model pretrained on COCO

2️ Replace Task-Specific Layers

For object detection:

Change the number of output classes

Modify classification head

Keep backbone features

Example:
COCO has 80 classes → your dataset may have 5 classes.

3️ Freeze or Unfreeze Layers

You can:

🔹 Freeze Backbone

Keep early convolution layers fixed

Train only detection head

Q29. What is the difference between semantic segmentation and instance segmentation?

Ans. Both tasks fall under image segmentation, but they differ in how objects are identified.

1️ Semantic Segmentation
Definition

Semantic segmentation assigns a class label to every pixel in an image.

Key characteristic:

All objects of the same class are treated as one region.

Example:

If there are 5 cars → all car pixels are labeled as “car”

No distinction between individual cars

Output

Pixel-wise classification map

One mask per class

Use Cases

Road scene understanding

Medical imaging

Satellite imagery

2️ Instance Segmentation
Definition

Instance segmentation also labels every pixel, but it distinguishes between individual objects of the same class.

Key characteristic:

Each object instance gets its own mask.

Example:

5 cars → 5 separate masks

Each car is uniquely identified

Output

Pixel-wise masks

Separate mask per object instance

Use Cases

Autonomous driving

Retail analytics

Robotics

 Core Difference
Feature	Semantic Segmentation	Instance Segmentation
Pixel labeling	Yes	Yes
Distinguishes object instances	❌ No	✅ Yes
Output type	Class map	Object-wise masks
Complexity	Lower	Higher
 Technical Perspective

Semantic segmentation solves a per-pixel classification problem.

Instance segmentation solves a detection + segmentation problem.

Instance segmentation requires:

Object localization

Mask generation per object

Models like Mask R-CNN are commonly used for instance segmentation.
